```python
# Praktikum 4 - Ekspresi slide 2
# Menghitung nilai akhir mahasiswa (5 komponen nilai, input via keyboard)

tugas      = float(input("Masukkan nilai tugas      : "))
kuis       = float(input("Masukkan nilai kuis       : "))
praktikum  = float(input("Masukkan nilai praktikum  : "))
uts        = float(input("Masukkan nilai UTS        : "))
uas        = float(input("Masukkan nilai UAS        : "))

nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)
standar_kelulusan = float(input("Masukkan standar kelulusan  : "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# Soal 3 - B1 (Pertemuan 3) (boleh mengemudi sendiri)
umur = int(input("Masukkan umur: "))
punya_sim = input("Punya SIM? (True/False): ").strip().lower() == "true"

boleh_mengemudi = umur >= 17 and punya_sim
print("Boleh mengemudi sendiri:", boleh_mengemudi)

# Latihan operator precedence:
# 'not' dikerjakan lebih dulu daripada 'and', jadi tanpa kurung hasilnya berbeda.
tidak_boleh_benar = not (umur >= 17 and punya_sim)   # pakai kurung (benar)
tidak_boleh_salah = not umur >= 17 and punya_sim      # tanpa kurung (salah)
print("Tidak boleh mengemudi (pakai kurung)  :", tidak_boleh_benar)
print("Tidak boleh mengemudi (tanpa kurung)  :", tidak_boleh_salah)
```

```python
# Soal 1 - Pertemuan 4

jumlah_buku = int(input("Jumlah buku      : "))
harga_buku = int(input("Harga per buku   : "))
diskon_persen = int(input("Diskon (%)       : "))
uang_bayar = int(input("Uang dibayarkan  : "))
jumlah_orang = int(input("Jumlah orang     : "))

total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon
kembalian = uang_bayar - bayar
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang
cukup = uang_bayar >= bayar

print()
print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)
```

```python
# Soal 2 - Pertemuan 4

jumlah_panel = int(input("Jumlah panel           : "))
daya_panel = int(input("Daya per panel (Wp)   : "))
jam_penyinaran = float(input("Penyinaran (jam/hari)  : "))
efisiensi = float(input("Efisiensi sistem (0-1) : "))
kebutuhan = float(input("Kebutuhan (kWh/hari)   : "))
tarif = float(input("Tarif (Rp/kWh)         : "))

# 1. Energi per hari (kWh)
energi_harian = jumlah_panel * daya_panel * jam_penyinaran * efisiensi / 1000

# 2. Energi per bulan (30 hari) dan nilai rupiahnya
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif

# 3. Kekurangan energi per hari
kekurangan = kebutuhan - energi_harian

# 4. Apakah PLTS mencukupi?
mencukupi = energi_harian >= kebutuhan

print()
print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
# Soal 3 - Pertemuan 4

daya = int(input("Daya generator (kW)     : "))
jam = int(input("Operasi (jam/hari)     : "))
konsumsi = float(input("Konsumsi (liter/kWh)   : "))
harga_solar = int(input("Harga solar (Rp/liter) : "))
kapasitas_tangki = int(input("Kapasitas tangki (L)   : "))
arus = int(input("Arus kabel (A)         : "))
hambatan = float(input("Hambatan kabel (ohm)   : "))

# 1. Energi per hari (kWh)
energi = daya * jam
# 2. Kebutuhan solar per hari dan biayanya
solar = energi * konsumsi
biaya = solar * harga_solar
# 3. Berapa hari penuh tangki cukup dan sisa solarnya
hari_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar
# 4. Rugi daya pada kabel (P = I^2 x R)
rugi_kabel = arus ** 2 * hambatan
# 5. Rugi tinggi: lebih dari 150 W dan arus minimal 20 A
rugi_tinggi = rugi_kabel > 150 and arus >= 20
print()
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
