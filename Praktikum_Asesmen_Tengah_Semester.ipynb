```python
# SOAL 9

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for i, nilai in enumerate(data, start=1):

    try:
        vibrasi = float(nilai)

    except ValueError:
        print(f"Unit {i}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if vibrasi > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")

    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break


print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

```python
# SOAL CHALLANGE BONUS

# PROGRAM PEMANTAUAN KONSUMSI LISTRIK HARIAN
# Masalah: Memantau penggunaan energi listrik selama 7 hari
# Tujuan: Menghitung total, rata-rata, dan mendeteksi
# penggunaan listrik yang tinggi.

konsumsi_listrik = []
print("=== PEMANTAUAN KONSUMSI LISTRIK 7 HARI ===")

for hari in range(1, 8):
    while True:
        try:
            energi = float(input(f"Masukkan konsumsi listrik hari ke-{hari} (kWh): "))

            if energi < 0:
                print("Konsumsi tidak boleh negatif!")
            else:
                konsumsi_listrik.append(energi)
                break

        except ValueError:
            print("Input harus berupa angka! Silakan coba lagi.")


total = sum(konsumsi_listrik)
rata_rata = total / len(konsumsi_listrik)

pri
for i in range(len(konsumsi_listrik)):
    energi = konsumsi_listrik[i]

    print(f"Hari ke-{i+1}: {energi} kWh")

    if energi > 15:
        print("Status: KONSUMSI TINGGI!")
    elif energi >= 10:
        print("Status: Konsumsi Normal")
    else:
        print("Status: Konsumsi Rendah")

print("\n=== RINGKASAN ===")
print(f"Total konsumsi listrik: {total:.2f} kWh")
print(f"Rata-rata konsumsi: {rata_rata:.2f} kWh/hari")

if rata_rata > 15:
    print("Peringatan: Rata-rata penggunaan listrik terlalu tinggi!")
elif rata_rata >= 10:
    print("Penggunaan listrik masih dalam kategori normal.")
else:
    print("Penggunaan listrik tergolong hemat.")

print("Pemantauan selesai.")
```
