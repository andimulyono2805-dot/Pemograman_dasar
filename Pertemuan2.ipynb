```python
# Latihan 1 Membuat Variable

tegangan = 18
arus = 6
tipe_sel = 'monocrystalline'

print(f"Tegangan: {tegangan} V")
print(f"Arus: {arus} A")
print(f"Tipe Sel: {tipe_sel}")
```

```python
# Latihan 2 Tipe primitif & type()

resistor_ohm = 1200
dioda_ma = 500.0
kapasitor = '15 μF'
sesuai_standar = True

print(f"Resistor: {resistor_ohm} {type(resistor_ohm)}")
print(f"Dioda: {dioda_ma} {type(dioda_ma)}")
print(f"kapasitor: {kapasitor} {type(kapasitor)}")
print(f"sesuai_standar: {sesuai_standar} {type(sesuai_standar)}")
```

```python
# Latihan 3 Memilih tipe yang tepat

Nama_unit = "PLTS Nunukan"
Jumlah_Panel = 240
Tegangan_Keluaran = 48.5
Sedang_beroperasi = True

print(f"Nama_Unit: {Nama_unit} {type(Nama_unit)}")
print(f"Jumlah_Panel: {Jumlah_Panel} {type(Jumlah_Panel)}")
print(f"Tegangan_Keluaran: {Tegangan_Keluaran} {type(Tegangan_Keluaran)}")
print(f"Sedang_Bekerja: {Sedang_beroperasi} {type(Sedang_beroperasi)}")
```

```python
# Latihan 4 List

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

print(f"nilai pertama: {tegangan_harian[0]} V")
print(f"nilai terakhir: {tegangan_harian[-1]} V")

tegangan_harian.append(222.0)
print(f"Daftar setelah penambahan: {tegangan_harian}")
print(f"Jumlah total data: {len(tegangan_harian)}")
```

```python
# Latihan 5 Dictionary

motor_spesifikasi = {
    'nama': 'gesits_G1',
    'daya_kW': 2.000,
    'tegangan_V': 72,
    'rpm': 3000
}
print(f"Nama motor: {motor_spesifikasi['nama']}")
motor_spesifikasi['rpm'] = 3500
print("\nSpesifikasi motor setelah perubahan:")
print(motor_spesifikasi)
```

```python
# Jawaban Latihan 6 Gabungan
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

print(f"Nama unit kedua: {pembangkit[1]['nama']}")

pembangkit[0]['status'] = 'standby'

pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "konstruksi"})

print("\nData pembangkit setelah perubahan:")
for unit in pembangkit:
    print(unit)
```
