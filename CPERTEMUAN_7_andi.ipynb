```python
#AKTIVITAS 6F


skor = [80, 65, 90, 75, 70]

total = 0
jumlah_lulus = 0

for nilai in skor:
    total = total + nilai
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)

print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
#AKTIVITAS 6G


suhu = [28, 31, 30, 33, 29, 27, 32]

rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")

#Dari 7 nilai suhu, diperoleh 3 rata-rata bergerak 5 hari, yaitu: 30,2°C 30,0°C 30,2°C
#Karena setiap rata-rata menggunakan 5 data suhu secara berurutan. Jadi, jumlah rata-rata yang diperoleh adalah: 7-5+1=3

```

```python
#AKTIVITAS 6H


energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

hari = list(range(1, 31))
jam_operasi = [24] * 30
jam_operasi[16] = 18

NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

# 1. Memeriksa penurunan energi harian
print("=== PENURUNAN ENERGI ===")

for i in range(1, len(energi)):
    perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100

    if perubahan < -20:
        print(f"Hari {i + 1}: {perubahan:.1f}%")

# 2. Memeriksa jam operasi
print("\n=== JAM OPERASI ===")

for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print("Hari:", hari[i])
        print("Jam operasi:", jam_operasi[i], "jam")
        print("Energi:", energi[i], "MWh")

# 3. Menghitung rata-rata hari kerja dan akhir pekan
kerja = []
libur = []

for i in range(len(energi)):
    if NAMA_HARI[i % 7] == "Sab" or NAMA_HARI[i % 7] == "Min":
        libur.append(energi[i])
    else:
        kerja.append(energi[i])

print("\n=== RATA-RATA ENERGI ===")
print("Hari kerja:", round(sum(kerja) / len(kerja)), "MWh")
print("Akhir pekan:", round(sum(libur) / len(libur)), "MWh")

# 4. Analisis kemungkinan penyebab
print("\n=== ANALISIS HARI KE-17 ===")

penurunan = (energi[16] - energi[15]) / energi[15] * 100

print("Energi hari ke-16:", energi[15], "MWh")
print("Energi hari ke-17:", energi[16], "MWh")
print("Penurunan energi:", round(abs(penurunan), 1), "%")
print("Jam operasi:", jam_operasi[16], "jam")

print("\nKemungkinan penyebab:")
print("1. Jam operasi berkurang.")
print("2. Mesin mengalami kerusakan atau maintenance.")
print("3. Beban produksi atau permintaan energi menurun.")
```

```python
#AKTIVITAS 6I


# Membuat list suhu selama 7 hari
suhu = [28, 31, 30, 33, 29, 27, 32]

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# 2. Menambahkan dan mengeluarkan satu data
suhu.append(34)
print("Setelah append:", suhu)

suhu.pop()
print("Setelah pop:", suhu)

# 3. Menampilkan suhu yang sudah diurutkan
print("Suhu terurut:", sorted(suhu))
print("Data asli:", suhu)

# 4. Menghitung rata-rata dan suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari_tertinggi = suhu.index(tertinggi) + 1

print("Rata-rata suhu:", rata_rata)
print("Suhu tertinggi:", tertinggi)
print("Suhu tertinggi terjadi pada hari ke-", hari_tertinggi)

# 5. Membuat list suhu di atas rata-rata
di_atas_rata = []

for nilai in suhu:
    if nilai > rata_rata:
        di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", di_atas_rata)
```
